```python
import json
import numpy as np
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
from tensorflow import keras
from TheModelFinalFinalV2 import build
from tqdm import tqdm
from tensorflow.keras.models import load_model
```

```python
#Datos generales para vlidar el modelo
_, (x_test, y_test) = keras.datasets.mnist.load_data()
x_test = np.expand_dims(x_test, -1) / 255.0  


#Cargar segmentos de cada miembro del equipo
local_data = []
n = 5  #Cantidad de partes (personas por equipo)
for i in range(n):
    with open(f"data_part_{i}.json") as f:
        data = json.load(f)
    x = np.array(data["x"]).astype(np.float32) / 255.0  
    x = np.expand_dims(x, -1)  
    y = np.array(data["y"])
    local_data.append((x, y))
```

```python
# Función para entrenar y validar el modelo
def train_and_verify(local_model, x_train_local, y_train_local, epochs, x_test, y_test, save_name):
    history = local_model.fit(x_train_local, y_train_local, epochs=epochs, validation_data=(x_test, y_test))
    y_pred = local_model.predict(x_test)
    y_pred_classes = np.argmax(y_pred, axis=1)
    print(classification_report(y_test, y_pred_classes))

    plt.plot(history.history['accuracy'], label='Train Acc')
    plt.plot(history.history['val_accuracy'], label='Val Acc')
    plt.legend()
    plt.title(save_name)
    plt.show()

    local_model.save(save_name)

for i in range(n):  
    x_train_local, y_train_local = local_data[i] 

    local_model = build.build_it()
    save_name = f'local_model_{i}.keras'
    train_and_verify(local_model, x_train_local, y_train_local, epochs=5, x_test=x_test, y_test=y_test, save_name=save_name)
```
